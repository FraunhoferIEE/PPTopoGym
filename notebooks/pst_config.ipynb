```python
cfg = config_30pst()
config_agent = copy.deepcopy(cfg)
env = PPTopoGym(cfg)
index = 0
obs, info = env.reset(options={"index": index})
actions = spaces.Discrete(len(cfg["action_space"]))
agent_greedy = GreedyAgent(actions, config_agent, n_workers=16)
overloads = []
for _ in range(3):
    action = agent_greedy.act(obs, info)
    obs, reward, terminated, truncated, info = env.step(action)
    overloads.append(round(float(env.net.res_line.loading_percent.max()),2))
print(overloads)
```
